**Попробуем сравнить мою реализацию ViT и реализацию из timm**

```python
timm_vit = timm.create_model(
    'vit_tiny_patch16_224',
    pretrained=False,
    num_classes=10
)
my_vit = ViT(
    image_size=224,
    patch_size=16,
    embed_dim=192,
    n_heads=3,
    n_blocks=12,
    mlp_ratio=4.0,
    num_classes=10
)
```

```python
def count_params(model):
    return sum(p.numel() for p in model.parameters())

print("Моя модель:  ", count_params(my_vit))
print("timm ViT-Tiny:", count_params(timm_vit))
```
